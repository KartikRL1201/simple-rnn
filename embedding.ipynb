```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=['The glass of milk',
'the glass of juice',
'the cup of tea',
'I am a good boy',
'I am a good developer',
'Understand the meaning of words',
'Your videos are good',]
```

```python
sent
```

```python
vocab_size=10000
one_hot_repr=[one_hot(words,vocab_size)for words in sent]
one_hot_repr
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_len=len(max(one_hot_repr,key=len))
embedded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_len)
print(embedded_docs)
```

```python
dim=10
model=Sequential()
model.add(Embedding(vocab_size,dim,input_length=sent_len))
model.compile('adam','mse')
model.summary()
```

```python
model.predict(embedded_docs)
```
